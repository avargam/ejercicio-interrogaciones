## Código solución (NO MODIFICAR)

```python
def armonizar(sinfonia):

    armonizador = sinfonia[0]
    armonizado = [sinfonia[0]]

    for i in range(1, len(sinfonia)):
        instrumento = sinfonia[i]
        pos = 0
        instrumento_armonizado = []
        for j in range(len(armonizador)):
            nota_armonizada = False
            if pos < len(instrumento):
                if instrumento[pos] == armonizador[j]:
                    instrumento_armonizado.append(instrumento[pos])
                    nota_armonizada = True
                    pos += 1
            if not nota_armonizada:
                instrumento_armonizado.append("-")
        armonizado.append(instrumento_armonizado)

    return armonizado
```
